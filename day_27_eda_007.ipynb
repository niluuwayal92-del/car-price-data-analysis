```python
import pandas as pd
```

```python
#load the dataset
df = pd.read_csv("/content/car_price_dataset.csv")
df
```

```python
#How many brands------->10
df['Brand'].unique().size
```

```python
df['Fuel_Type'].unique()
```

```python
#1. What is the average price of all cars in the dataset?
df['Price'].mean()
print(f"the average price of all cars in the dataset{df['Price'].mean()}")
```

```python
#2. Which car brand has the highest average price?
b=df.groupby("Brand")["Price"].mean().idxmax()
print(f"the car brand with the higher average priceis{b}")
```

```python
#3. What is the most common fuel type in the dataset?
f=df['Fuel_Type'].value_counts().idxmax()
print(f"the most common fuel type in the dataset is{f}")
#
```

```python
#4. How many cars have a mileage greater than 200,000?
b=df['Transmission'].value_counts()
print(f"The cars have a mileage greater than 200,000is{b}")
```

```python
#5. What is the distribution of cars by transmission type?
c=df['Transmission'].value_counts()
print(f"The distribution of cars by transmission type is{c}")
```

```python
#6. Which car model has the highest price?
g=df['Price'].max()
print(f"The car model has highest price is{g}")
```

```python
#7. What is the average price of cars by fuel type?
m=df.groupby("Fuel_Type")['Price'].mean()
print(f"The average price of car by fuel type is{m}")
```

```python
#8. How many cars are from the year 2020 or later?
t=df[df['Year']>=2020].shape[0]
print(f"The cars are from the year 2020 or later is{t}")
```

```python
#9. What is the average engine size for each brand?
x=df.groupby("Brand")["Engine_Size"].mean()
print(f"The average engine size for each brand is{x}")
```

```python
#10. Which car has the lowest mileage?
z=df[df['Mileage']==df['Mileage'].min()]
print(f"the car has the lowest mileage is{z}")
```

```python
#11. What is the correlation between mileage and price?
s=df['Mileage'].corr(df['Price'])
print(f"The correlation between mileage and price{s}")
```

```python
#12. How many cars have more than 2 owners?
n=(df['Owner_Count'] >2).sum()
print(f"the cars have more than 2 owners is{n}")
```

```python
#13. What is the average price of cars by year?
b=(df.groupby('Year')['Price'].mean())
print(f"The average price of cars by year{b}")
```

```python
#14. Which car has the highest number of doors?
max_doors=df['Doors'].max()
l=df[df['Doors'] == max_doors]
print(f"The car has higest number of doors is{l}")
```

```python
#15. What is the average mileage for each fuel type?
t=df.groupby('Fuel_Type')['Mileage'].mean()
print(f"The average mileage for each fuel type is{t}")
```

```python
#16. How many cars are from the brand 'Toyota'?
u=(df['Brand'] == 'Toyota').sum()
print(f"the cars are from the brand 'Toyota is{u}")
```

```python
#17. What is the average price of cars with automatic transmission?
k=(df[df['Transmission'] == 'Automatic']['Price'].mean())
print(f"The average price of cars with automatic transmission is{k}")
```

```python
#18. Which car has the highest owner count?
max_owner=df['Owner_Count'].max()
y=(df[df['Owner_Count'] == max_owner])
print(f"the car has the highest owner count is{y}")
```

```python
#19. What is the average engine size for cars with diesel fuel type?
j=(df[df['Fuel_Type'] == 'Diesel']['Engine_Size'].mean())
print(f"the average engine size for cars with diesel fuel type is{j}")
```

```python
#20. How many cars have a price greater than $10,000?
h=(df['Price'] > 10000).sum()
print(f"the car have a price greater than $10,000 is {h}")
```
